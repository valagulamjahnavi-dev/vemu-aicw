```python
#step1:import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#step:2 identifying the datasets
rdf=pd.read_csv("C:\\Users\\valag\\Downloads//bank-selected-columns.csv")
```

```python
#step 6: define features and target

x=rdf[["age"]]

y=rdf[["balance"]]
```

```python
#step5: visualization of datasets
import matplotlib.pyplot as plt
plt.figure(figsize=(3,4))
plt.scatter(rdf["age"],rdf["balance"],color="red",s=60)
plt.show()
```

```python
#step 7: splitting dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=42)

#interpretation: 80% training data, 20% testing data
```

```python
#step 8: model creation: Linear regression

from sklearn.linear_model import LinearRegression
lrm=LinearRegression()
fitted_lrm=lrm.fit(x_train, y_train)
```

```python
#step 10
y_pre=fitted_lrm.predict(x_test)
```

```python
# step 9:Identifying Model parameters
print("Slope:",fitted_lrm.coef_)
print("Intercept:",fitted_lrm.intercept_)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#step 12:Model Evaluattion
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np
mae=mean_absolute_error(y_test,y_pre)
mse=mean_squared_error(y_test,y_pre)
smse=np.sqrt(mse)
r2=r2_score(y_test,y_pre)
print(mae)
print(mse)
print(smse)
print(r2)
```
